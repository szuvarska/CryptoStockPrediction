```python
def unpack_results(df,results_col='results'):
    """
    Rozpakowuje zagnieżdżoną kolumnę wyników (np. z API rynkowych) do płaskiej struktury tabelarycznej.

    Funkcja:
    - pobiera pierwszy element listy w kolumnie `results`
    - wyciąga pola OHLCV (open, high, low, close, volume, adjusted_close)
    - konwertuje znacznik czasu w milisekundach na typ `timestamp`
    - usuwa dane wejściowe i kolumny pomocnicze

    Parametry:
        df (DataFrame): wejściowy Spark DataFrame zawierający kolumnę z wynikami
        results_col (str): nazwa kolumny zawierającej listę wyników (domyślnie "results")

    Zwraca:
        DataFrame: przetworzony DataFrame z płaskimi kolumnami cenowymi i timestampem
    """
    df = df.withColumn("res", F.col(results_col).getItem(0))
    
    df = df.withColumn("interval", F.col("res.v")) \
           .withColumn("open", F.col("res.o")) \
           .withColumn("high", F.col("res.h")) \
           .withColumn("low", F.col("res.l")) \
           .withColumn("close", F.col("res.c")) \
           .withColumn("adjusted_close", F.col("res.vw")) \
           .withColumn("timestamp", F.col("res.t")) \
           .withColumn("volume", F.col("res.n"))
    
    res = df.withColumn("datetime", F.from_unixtime(F.col("timestamp") / 1000).cast("timestamp"))
    res = res.drop("timestamp", 'results', 'res')
    return res
def split_ticker(df, ticker_col="ticker"):
    """
    Tworzy kolumny currency_from i currency_to z kolumny ticker np. 'C:USDAED' na 'USD' i 'AED'
    """
    df = df.withColumn("ticker_clean", F.regexp_replace(F.col(ticker_col), "^C:", ""))
    df = df.withColumn("currency_from", F.col("ticker_clean").substr(1,3))
    df = df.withColumn("currency_to", F.expr("substring(ticker_clean, 4, length(ticker_clean))"))
    df = df.drop("ticker_clean")
    return df
```

```python
unpacked = unpack_results(df2)
unpacked.show(vertical=True, truncate=False)
```

```python
currency_splitted = split_ticker(unpacked)
currency_splitted.show(vertical=True, truncate=False)
```

# Preprocessing i zapis do csv

```python
df = spark8.read.parquet("hdfs://namenode:8020/nifi/forex/")
```

```python
df = split_ticker(df)
df = unpack_results(df)
```

```python
df.write \
    .option("header", "true") \
    .mode("overwrite") \
    .csv("hdfs://namenode:8020/ready_data/forex")
```
