```python
from pyspark.sql import SparkSession
import pyspark
print(pyspark.__version__)
```

```python

# Przykładowe połączenie Sparka z Hive'm:
# appName, master (tak jak normalnie)
# Hive:
# .enableHiveSupport() - spark sql zaczyna korzystać domyślnie z Hive (wydaje mi się pobiera konfigurację z /opt/spark/conf/hive-site.xml)
# .config("hive.metastore.uris", "thrift://hive-metastore:9083") - mówi mu gdzie jest połączenie do metastore'a (w sumie chyba można byłoby to zhardkodować w hive-site.xml, ale wolę już nie psuć)
spark = SparkSession.builder \
    .appName("test-hive2") \
    .master("spark://spark-master:7077") \
    .enableHiveSupport() \
    .config("hive.metastore.uris", "thrift://hive-metastore:9083") \
    .getOrCreate()

# Wyświetl domyślną ścieżkę w których będą tworzone nowe bazy danych
print(spark.conf.get("spark.sql.warehouse.dir"))

# Pokaż bazy danych Hive
print("---------Bazy Danych-------------")
spark.sql("SHOW DATABASES").show()

# Zobaczmy metadane bazy testdb
print("---------Metadane testdb-------------")
spark.sql("DESCRIBE DATABASE testdb").show(truncate=False)

# Korzystamy z przykładowej bazy danych
spark.sql("USE testdb")

# Odczytujemy dane
print("----------SELECT z testdb.users------------")
df = spark.sql("SELECT * FROM users")
df.show()

# Sprawdzamy opis tabeli
print("---------Metadane testdb.users-------------")
spark.sql("DESCRIBE FORMATTED users").show(truncate=False)
```
