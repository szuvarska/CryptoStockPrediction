```python
from pyspark.sql import SparkSession
```

```python
df2 = spark.read.parquet("hdfs://namenode:8020/nifi/crypto-prices/crypto-quotes-20251106-214622.parquet")
df2.show(5)
```

```python
df3 = spark.read.parquet("hdfs://namenode:8020/nifi/stock-prices/stock_prices_2025-11-15-11.parquet")

for i, row in enumerate(df3.take(7)):
    print(f"Row: {i}")
    for col, val in row.asDict().items():
        print(f"{col:30s} {val}")
    print(f"{'-'*50}")
```
