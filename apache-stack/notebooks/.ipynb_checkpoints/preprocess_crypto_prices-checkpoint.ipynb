```python
from pyspark.sql import SparkSession
from pyspark.sql.utils import AnalysisException
```

```python
spark = SparkSession.builder \
    .appName("HDFS_test") \
    .master("spark://spark-master:7077") \
    .config("spark.sql.warehouse.dir", "hdfs://namenode:8020/user/nifi/crypto-prices") \
    .getOrCreate()
```

```python
df2 = spark.read.parquet("hdfs://namenode:8020/nifi/crypto-prices/crypto-quotes-20251106-214622.parquet")
```

```python
df2.show(1000)
```

```python
from pyspark.sql import functions as F
def get_symbol_clear(df):
    res = df.withColumn(
    "coin",
    F.split("symbol", ":")[1].substr(1, 3))
    return res
```

```python
get_symbol_clear(df2).show(5)
```

```python
def get_mean_cost(df):
    res = df.withColumn(
    "mean_price",
    (F.col('high')+F.col('low'))/2)
    return res
```

```python
get_mean_cost(df2).show(5)
```

```python
def get_datetime(df):
    res = df.withColumn(
    "datetime",
    F.from_unixtime(F.col("timestamp")).cast("timestamp"))
    return res
```

```python
get_datetime(df2).show(5)
```

# Preprocessing całego folderu i zapis do csv

```python
df = spark.read.parquet("hdfs://namenode:8020/nifi/crypto-prices/")
```

```python
df = get_symbol_clear(df)
df = get_datetime(df)
df = get_mean_cost(df)
```

```python
df.write \
    .option("header", "true") \
    .mode("overwrite") \
    .csv("hdfs://namenode:8020/ready_data/crypto-prices")
```
