```python
import os
import sys
import pyspark
import pkg_resources

print("="*40)
print("🔎 JUPYTER ENVIRONMENT DIAGNOSTICS")
print("="*40)

# 1. PYTHON INTERPRETER
# This is the most critical check. Is it using /usr/bin/python or a Conda env?
print(f"\n[1] Python Executable used by Jupyter:\n{sys.executable}")

# 2. CRITICAL ENV VARIABLES
# Spark needs these. If JAVA_HOME is set here, copy it to your run_app.sh.
print("\n[2] Environment Variables:")
vars_to_check = ['JAVA_HOME', 'SPARK_HOME', 'HADOOP_HOME', 'PYTHONPATH', 'PYSPARK_PYTHON']
for v in vars_to_check:
    val = os.environ.get(v, "❌ Not Set")
    print(f"  {v:<15}: {val}")

# 3. PATH CHECK
# If JAVA_HOME isn't set, Spark finds 'java' on the PATH.
print("\n[3] Java on PATH:")
try:
    # This runs a shell command to find where 'java' really is
    if os.system("which java") != 0:
        print("  'java' command not found in PATH")
except Exception as e:
    print(f"  Error checking path: {e}")

# 4. INSTALLED PACKAGES
# Check versions of key libraries to ensure your venv matches.
print("\n[4] Key Package Versions:")
target_pkgs = ['pyspark', 'pandas', 'jinja2', 'shiny', 'py4j']
installed = {pkg.key: pkg.version for pkg in pkg_resources.working_set}

for p in target_pkgs:
    if p in installed:
        print(f"  {p:<15} == {installed[p]}")
    else:
        print(f"  {p:<15} == ❌ Not Installed")

print("="*40)
```
